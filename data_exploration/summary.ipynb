```python
# Notebook 6: Summary Report
import os
import numpy as np

# Path to data
SAVE_DIR = "../processed_data/"

# Load data
X_train = np.load(os.path.join(SAVE_DIR, 'X_train.npy'))
y_train = np.load(os.path.join(SAVE_DIR, 'y_train.npy'))
X_val = np.load(os.path.join(SAVE_DIR, 'X_val.npy'))
y_val = np.load(os.path.join(SAVE_DIR, 'y_val.npy'))
X_test = np.load(os.path.join(SAVE_DIR, 'X_test.npy'))
y_test = np.load(os.path.join(SAVE_DIR, 'y_test.npy'))

X_all = np.concatenate([X_train, X_val, X_test])
y_all = np.concatenate([y_train, y_val, y_test])

print("="*70)
print("DATA EXPLORATION SUMMARY REPORT")
print("="*70)

# Dataset overview
benign_total = np.sum(y_all == 0)
malignant_total = np.sum(y_all == 1)
total = len(y_all)

print(f"""
DATASET OVERVIEW:
-----------------
Total Images:      {total}
Training Images:   {X_train.shape[0]}
Validation Images: {X_val.shape[0]}
Test Images:       {X_test.shape[0]}
Image Size:        {X_train.shape[1]}x{X_train.shape[2]} pixels
Channels:          1 (grayscale)

CLASS DISTRIBUTION:
-------------------
Benign:    {benign_total} ({benign_total/total*100:.1f}%)
Malignant: {malignant_total} ({malignant_total/total*100:.1f}%)
Ratio:     {benign_total/malignant_total:.2f}:1

PIXEL STATISTICS (Combined):
----------------------------
Min:  {X_all.min():.3f}
Max:  {X_all.max():.3f}
Mean: {X_all.mean():.3f}
Std:  {X_all.std():.3f}
Median: {np.median(X_all):.3f}

KEY OBSERVATIONS:
-----------------
1. The dataset has {benign_total/malignant_total:.2f}x more benign than malignant images.
2. Images are pre-processed to {X_train.shape[1]}x{X_train.shape[2]} pixels.
3. All values appear to be normalized (z-score).
4. Training/Validation/Test splits are: {X_train.shape[0]}/{X_val.shape[0]}/{X_test.shape[0]}

DATA QUALITY:
-------------
NaN values present: {np.isnan(X_all).any()}
Infinite values present: {np.isinf(X_all).any()}

RECOMMENDATIONS:
----------------
1. Imbalance ratio is {benign_total/malignant_total:.2f}:1 
   - Consider using class weights or focal loss
2. Data is properly normalized - ready for modeling
3. No NaN or infinite values detected - data is clean
4. Consider augmentation for the minority class (malignant)
5. Recommended augmentation techniques:
   - Rotation (+/- 20 degrees) for breast positioning
   - Horizontal flip for symmetry
   - Zoom (0.8-1.2) for different breast sizes
   - Brightness adjustment for scan variations
""")

print("="*70)
print("All exploration notebooks complete.")
print("Check the data_exploration folder for visualizations.")
print("="*70)
```
